```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split

tf.random.set_seed(42)
np.random.seed(42)
```

# <font color='green'><u><b>Parte 1 - Conjunto de Dados</b></u></font>

Não altere o conteúdo da célula a seguir!

```python
# Carregando o dataset CIFAR-10
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Ajustando dimensões dos rótulos
y_train = y_train.reshape(-1,)
y_test = y_test.reshape(-1,)

# Separando validação do treino
x_train, x_val, y_train, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

# Classes do dataset
class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]
```

O dataset mencionado foi criado para treinar e testar redes neurais na identificação automática de objetos em imagens. O conjunto possui 60 mil imagens coloridas no formato RGB e as mesmas estão divididas em 10 classes diferentes. Entre os objetos presentes estão aviões, carros, pássaros, gatos, cachorros, cavalos, navios e caminhões. O dataset é bastante utilizado em estudos e aplicações de visão computacional e redes neurais convolucionais devido ao fato de que as imagens são pequenas e possuem diferentes fundos, posições e iluminações.

Cada imagem possui três dimensões: largura, altura e canais de cor. As dimensões do tensor (32, 32, 3) significa que a imagem possui 32 pixels de altura, 32 pixels de largura e 3 canais de cor. Esses canais representam as cores RGB, que são responsáveis pela composição das imagens coloridas, cada pixel possui um valor associado em cada um dos três canais, permitindo representar diferentes cores. Em imagens em tons de cinza, normalmente existe apenas um canal, já que cada pixel representa apenas intensidade luminosa. Em redes neurais convolucionais, imagens coloridas fornecem mais informações para o modelo aprender padrões visuais, enquanto imagens em tons de cinza possuem menos complexidade e menor quantidade de informações.

# <font color='green'><u><b>Parte 2 - MLP</b></u></font>

### <font color='#2D9CDB'>Q7) As redes neurais do tipo Multilayer Perceptron (MLP) utilizam camadas densamente conectadas e, portanto, esperam vetores unidimensionais como entrada. Explique por que é necessário transformar as imagens do CIFAR-10 antes de utilizá-las em uma MLP e discuta o que acontece com a estrutura espacial das imagens durante esse processo. Em seguida, construa uma rede neural MLP utilizando a API Sequential do Keras com uma camada Flatten, duas camadas densas escondidas com função de ativação ReLU e uma camada de saída com 10 neurônios e ativação Softmax.</font>

```python
mlp_model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(32,32,3)),

    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dense(256, activation='relu'),

    tf.keras.layers.Dense(10, activation='softmax')
])

mlp_model
```

As redes MLP esperam vetores unidimensionais como entrada, porém as imagens possuem formato tridimensional. Por esse motivo, torna-se necessário transformar cada imagem em um vetor unidimensional utilizando a camada Flatten. Durante esse processo, a estrutura espacial da imagem é perdida, pois pixels vizinhos deixam de manter suas relações de proximidade. Isso causa com que a MLP tenha mais dificuldade para aprender padrões espaciais importantes presentes nas imagens.

### <font color='#2D9CDB'>Q8) Exiba o resumo da arquitetura da rede neural criada utilizando o método summary() e analise a quantidade de parâmetros treináveis da MLP. Com base nas dimensões das imagens do CIFAR-10 e no número de neurônios utilizados nas camadas densas, discuta por que MLPs podem rapidamente se tornar modelos muito grandes para problemas de classificação de imagens.</font>

```python
mlp_model.summary()
```

Como as imagens do CIFAR-10 possuem 32×32×3 = 3072 entradas, o número de conexões cresce rapidamente nas camadas densas. Isso aumenta significativamente o custo computacional, o consumo de memória e o risco de overfitting. Em problemas de visão computacional, esse crescimento excessivo de parâmetros torna as MLPs menos eficientes quando comparadas às CNNs.

### <font color='#2D9CDB'>Q9) Compile a MLP utilizando o otimizador Adam com taxa de aprendizado de $10^{-3}$, a função de custo sparse_categorical_crossentropy e a métrica accuracy. Em seguida, explique o papel da função de custo durante o treinamento, o objetivo do algoritmo de otimização e por que a combinação entre Softmax e entropia cruzada é adequada para problemas de classificação multiclasse.</font>

```python
mlp_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

A função de custo é responsavél por medir o erro entre as previsões realizadas pela rede neural e os valores reais do dataset. Durante o treinamento, o objetivo do algoritmo de otimização é ajustar os pesos da rede para minimizar esse erro. O otimizador Adam combina eficiência computacional e adaptação automática da taxa de aprendizado. A combinação entre Softmax e entropia cruzada é boa para classificação multiclasse porque a Softmax transforma as saídas da rede em probabilidades, enquanto a entropia cruzada mede o quão próximas essas probabilidades estão das classes corretas.

### <font color='#2D9CDB'>Q10) Treine a rede neural por 25 épocas utilizando os conjuntos de treinamento e validação. Armazene o histórico do treinamento em um objeto do tipo History e, em seguida, plote gráficos mostrando a evolução da loss e da accuracy ao longo das épocas para os conjuntos de treinamento e validação. Após gerar os gráficos, analise o comportamento do treinamento, discutindo se o modelo aparenta convergir adequadamente, se há estabilidade durante o treinamento e se existem indícios de overfitting.</font>

```python
history_mlp = mlp_model.fit(
    x_train,
    y_train,
    epochs=25,
    validation_data=(x_val, y_val)
)

plt.figure(figsize=(12,5))

# Loss
plt.subplot(1,2,1)
plt.plot(history_mlp.history['loss'], label='Treino')
plt.plot(history_mlp.history['val_loss'], label='Validação')
plt.title('Loss')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

# Accuracy
plt.subplot(1,2,2)
plt.plot(history_mlp.history['accuracy'], label='Treino')
plt.plot(history_mlp.history['val_accuracy'], label='Validação')
plt.title('Accuracy')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

Durante o treinamento, espera-se que a loss diminua e a accuracy aumente gradualmente. Porém, se a accuracy de treinamento continuar aumentando enquanto a accuracy de validação para de melhorar ou começa a diminuir, isso indica overfitting. Em MLPs aplicadas é comum observar diferenças entre treino e validação devido à dificuldade da rede em generalizar padrões espaciais das imagens.

### <font color='#2D9CDB'>Q11) Avalie o desempenho da MLP utilizando o conjunto de teste e reporte os valores de loss e accuracy obtidos. Em seguida, compare os resultados dos conjuntos de treinamento e teste e discuta se o modelo aparenta apresentar boa capacidade de generalização.</font>

```python
test_loss, test_acc = mlp_model.evaluate(x_test, y_test)

print("Loss no teste:", test_loss)
print("Accuracy no teste:", test_acc)
```

A comparação entre os resultados de treinamento e teste permite avaliar a capacidade de generalização do modelo. Se a accuracy de teste for próxima da accuracy de treinamento, o modelo generaliza relativamente bem. Caso exista uma diferença muito grande, isso pode indicar overfitting.

### <font color='#2D9CDB'>Q12) Utilize a MLP treinada para gerar previsões sobre algumas imagens do conjunto de teste. Quando as probabilidades ficam muito distribuídas entre várias classes, significa que o modelo está menos confiante na decisão.

Isso acaba acontecendo porque a MLP perde a estrutura espacial das imagens ao utilizar Flatten, causando dificuldade na identificação de padrões locais importantes para distinguir objetos parecidos.

# <font color='green'><u><b>Parte 3 - CNN</b></u></font>

### <font color='#2D9CDB'>Q15) As redes neurais convolucionais (CNNs) foram desenvolvidas para lidar melhor com dados que possuem estrutura espacial, como imagens. Explique por que arquiteturas MLP apresentam limitações em problemas de visão computacional e descreva como as CNNs conseguem preservar relações espaciais entre os pixels. Em seguida, explique os conceitos de convolução, filtros convolucionais e mapas de características (feature maps).</font>

As MLPs apresentam limitações em visão computacional porque não preservam a estrutura espacial das imagens. Quando a imagem é achatada em um vetor, relações importantes entre pixels vizinhos são perdidas. As CNNs foram desenvolvidas para resolver esse problema, mantendo a organização espacial das imagens e aprendendo padrões locais automaticamente. A convolução consiste na aplicação de filtros sobre a imagem para detectar características importantes, como bordas, texturas e formas. Esses filtros convolucionais percorrem a imagem gerando mapas de características (feature maps), que representam regiões onde determinados padrões foram encontrados.

### <font color='#2D9CDB'>Q16) Construa uma rede neural convolucional utilizando a API Sequential do Keras. A arquitetura da CNN deve conter camadas Conv2D com ativação ReLU, camadas MaxPooling2D, uma camada Flatten e uma camada de saída com 10 neurônios e ativação Softmax. Após construir a arquitetura, explique o papel das camadas convolucionais, das camadas de pooling e da camada Flatten dentro da CNN.</font>

```python
cnn_model = tf.keras.Sequential([

    tf.keras.layers.Conv2D(
        32,
        (3,3),
        activation='relu',
        input_shape=(32,32,3)
    ),

    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Conv2D(
        64,
        (3,3),
        activation='relu'
    ),

    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Conv2D(
        64,
        (3,3),
        activation='relu'
    ),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(64, activation='relu'),

    tf.keras.layers.Dense(10, activation='softmax')
])
```

As camadas convolucionais extraem características importantes das imagens utilizando filtros aprendidos automaticamente. As camadas de MaxPooling reduzem a dimensionalidade dos mapas de características, diminuindo o custo computacional e ajudando na generalização. A camada Flatten transforma os mapas de características em um vetor unidimensional para permitir a classificação final pelas camadas densas.

Em seguida, discuta por que CNNs costumam ser mais eficientes para classificação de imagens, mesmo utilizando menos parâmetros treináveis.</font>

```python
cnn_model.summary()
```

As CNNs normalmente possuem menos parâmetros treináveis que MLPs porque utilizam compartilhamento de pesos nos filtros convolucionais. Isso reduz significativamente o número de conexões necessárias. Além disso, CNNs conseguem aprender padrões visuais locais de maneira mais eficiente, tornando-se mais adequadas para classificação de imagens mesmo com arquiteturas menores.

```python
history_cnn = cnn_model.fit(
    x_train,
    y_train,
    epochs=25,
    validation_data=(x_val, y_val)
)
```

```python
plt.figure(figsize=(12,5))

# Loss
plt.subplot(1,2,1)
plt.plot(history_cnn.history['loss'], label='Treino')
plt.plot(history_cnn.history['val_loss'], label='Validação')
plt.title('Loss CNN')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

# Accuracy
plt.subplot(1,2,2)
plt.plot(history_cnn.history['accuracy'], label='Treino')
plt.plot(history_cnn.history['val_accuracy'], label='Validação')
plt.title('Accuracy CNN')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

A CNN tende a apresentar convergência mais estável e accuracy superior à MLP. Isso ocorre porque as convoluções conseguem aprender padrões espaciais importantes das imagens. Normalmente, observa-se menor diferença entre treino e validação, indicando melhor capacidade de generalização.
